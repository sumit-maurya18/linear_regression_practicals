```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
%matplotlib inline
```

```python
x = 6 * np.random.rand(100, 1) - 3
y = 0.5 * x**2 + 1.5 * x + 2 + np.random.rand(100, 1)
#quadratic eqn used : y = 0.5x^2 + 1.5x + 2 + outliers
```

```python
plt.scatter(x,y, color='red')
plt.xlabel('X-Dataset')
plt.ylabel('Y-Label')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
#implement simple linear regression
from sklearn.linear_model import LinearRegression
simple_regression = LinearRegression()
```

```python
simple_regression.fit(x_train, y_train)
```

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(y_test, simple_regression.predict(x_test))
print(score)
```

```python
#visualize
plt.plot(x_train, simple_regression.predict(x_train), color='r')
plt.scatter(x_train, y_train)
plt.xlabel('X-dataset')
plt.ylabel('Y-datset')
plt.show()
```

```python
#using polynomial regression
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=True)
```

```python
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
```

```python
x_train_poly
```

```python
x_test_poly
```

```python
regression = LinearRegression()
regression.fit(x_train_poly, y_train)
```

```python
y_predicted_poly = regression.predict(x_test_poly)
```

```python
score_poly = r2_score(y_test, y_predicted_poly)
print(score_poly)
```

```python
regression.coef_
```

```python
regression.intercept_
```

```python
plt.scatter(x_train, regression.predict(x_train_poly))
plt.scatter(x_train, y_train)
plt.show()
```

Changing the degree to 3

```python
#using polynomial regression and now changing degree to 3
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=3, include_bias=True)
```

```python
y_predicted_poly = regression.predict(x_test_poly)
```

Prediction of new data

```python
x_new = np.linspace(-3, 3, 200).reshape(200,1)
x_new_poly = poly.transform(x_new)
```

```python
x_new_poly
```

```python
y_new = regression.predict(x_new_poly)
```

```python
plt.plot(x_new, y_new, 'r', label = 'Predictions')
plt.plot(x_train, y_train, 'b', label = 'Training Points')
plt.plot(x_test, y_test, 'g', label='Testing Points')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
