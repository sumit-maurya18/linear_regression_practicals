```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5.0, 5.0, 100)
y = np.sqrt(10**2 - x **2)
y = np.hstack([y, -y])
x = np.hstack([x, -x])
```

```python
x
```

```python
y
```

```python
x1 = np.linspace(-5.0, 5.0, 100)
y1 = np.sqrt(5**2, x1**2)
y1 = np.hstack([y1, -y1])
x1 = np.hstack([x1, -x1])
```

```python
plt.scatter(y, x)
plt.scatter(y1, x1)
```

```python
import pandas as pd
df1 = pd.DataFrame(np.vstack([y,x]).T, columns=['X1', 'X2'])
df1['Y'] = 0
df2 = pd.DataFrame(np.vstack([y1, x1]).T, columns=['X1', 'X2'])
df2['Y'] = 1
df = pd.concat([df1, df2], ignore_index=True)
df.head(5)
```

```python
df.tail(5)
```

```python
##Creating dependent and independent features
X = df.iloc[:, :2]
Y=df.Y
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split

x_train, y_train, y_test, y_train = train_test_split(X, Y, test_size = 0.25, random_state = 10)
```

```python
x_train
```

```python
y_train
```

```python
##Applyimmg polynomial kernel trannsforamtion
##find component for polynomial kernel
#x1, x2, x1_sq, x2_sq, x1*x2
df['X1_square'] = df['X1']**2
df['X2_square'] = df['X2']**2
df['X1 * X2'] = (df['X1'] * df['X2'])
df.head()
```

```python
##independent and dependent features
x = df[['X1', 'X2', 'X1_square', 'X2_square', 'X1 * X2']]
y = df['Y']
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
```

```python
x_train
```

```python
import plotly.express as px

fig = px.scatter_3d(df, x = 'X1', y = 'X2', z = 'X1 * X2', color = Y)
fig.show()
```

```python
fig = px.scatter_3d(df, x = 'X1_square', y = 'X2_square', z = 'X1 * X2', color = Y)
fig.show()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

classifier = SVC(kernel = 'linear')
classifier.fit(x_train, y_train)
y_pred = classifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
classifier = SVC(kernel = 'poly') #without manually creating features
classifier.fit(x_train, y_train)
y_pred = classifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
classifier = SVC(kernel = 'rbf') #without manually creating features
classifier.fit(x_train, y_train)
y_pred = classifier.predict(x_test)
accuracy_score(y_test, y_pred)
```
